```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/CO2_Emissions_Canada.csv")
target = "Fuel Consumption Comb (L/100 km)"

df.info()
print(df.describe())
print("Missing:", df.isna().sum().sum(), "| Duplicates:", df.duplicated().sum())

sns.histplot(df[target], kde=True); plt.show()                       # target distribution
sns.scatterplot(data=df, x="Engine Size(L)", y=target); plt.show()   # engine vs consumption
sns.boxplot(data=df, x="Fuel Type", y=target); plt.show()            # fuel type
sns.boxplot(data=df, x="Cylinders", y=target); plt.show()            # cylinders
plt.figure(figsize=(10, 4))
sns.boxplot(data=df, x="Vehicle Class", y=target)
plt.xticks(rotation=75); plt.show()                                  # vehicle class
sns.heatmap(df.select_dtypes("number").corr(), annot=True, cmap="coolwarm"); plt.show()
```
